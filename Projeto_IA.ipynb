## Análise Exploratória do Dataset de exoplanetas da Missão Kepler.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Carregamento do arquivo Excel
df = pd.read_excel('/content/dataset_exo.xlsx')

# 1. Filtro de Colunas Relevantes para o Revisionismo
cols_ia = ['koi_disposition', 'koi_period', 'koi_duration', 'koi_depth', 'koi_model_snr']
df_audit = df[cols_ia].dropna()

# 2. Visualização das Classes (Alvo da Replicação)
plt.figure(figsize=(8, 5))
sns.countplot(data=df_audit, x='koi_disposition', palette='viridis')
plt.title('Distribuição de Classes no Catálogo Kepler')
plt.show()

# 3. Análise de SNR do Modelo vs. Profundidade
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df_audit, x='koi_depth', y='koi_model_snr', hue='koi_disposition', alpha=0.4)
plt.xscale('log')
plt.yscale('log')
plt.title('Relação Profundidade vs koi_model_snr')
plt.grid(True, which="both", ls="-", alpha=0.2)
plt.show()

# 4. Boxplot para Identificação de Assinaturas Físicas
plt.figure(figsize=(10, 6))
sns.boxplot(data=df_audit, x='koi_disposition', y='koi_model_snr')
plt.yscale('log')
plt.title('Distribuição de SNR por Classificação')
plt.show()
```
